````python
def train_test_split(X, y, test_size=0.2, random_state=None):
    if random_state is not None:
        np.random.seed(random_state)
    indices = np.arange(len(X))
    np.random.shuffle(indices)
    test_size = int(len(X) * test_size)
    test_indices = indices[:test_size]
    train_indices = indices[test_size:]
    return X[train_indices], X[test_indices], y[train_indices], y[test_indices]
````

### Soal A5 – Gradient Descent & Learning Rate
1. Latih model `LinearRegressionGD` (dari materi) dengan `n_iterations=5000` menggunakan **3 nilai learning rate**: `0.001`, `0.02`, dan `0.1`.
2. Untuk setiap learning rate, tampilkan intercept dan koefisien akhirnya, lalu bandingkan dengan hasil normal equation (Soal A3).
3. Buat grafik **loss (MSE) vs iterasi** untuk ketiga learning rate (boleh dalam satu grafik atau *subplot*; gunakan skala log jika perlu).
4. Jelaskan apa yang terjadi pada masing-masing learning rate: mana yang **terlalu lambat**, mana yang **pas**, dan mana yang **gagal konvergen (divergen)**? Mengapa bisa terjadi?

> *Tips: jika muncul `RuntimeWarning: overflow`, itu adalah bagian dari hasil eksperimen. Jelaskan artinya.*

````python
# Jawaban A5 - Gradient Descent dengan Berbagai Learning Rate

# 1. Definisikan Kelas LinearRegressionGD manual
class LinearRegressionGD:
    def __init__(self, n_iterations=5000, lr=0.01):
        self.n_iterations = n_iterations
        self.lr = lr
        self.theta = None
        self.loss_history = []

    def fit(self, X, y):
        m_samples = len(X)
        # Tambahkan kolom bias (1) ke X_train
        X_b = np.c_[np.ones((m_samples, 1)), X]
        self.theta = np.zeros(X_b.shape[1])
        self.loss_history = []

        for i in range(self.n_iterations):
            # Hitung prediksi (y_pred)
            y_pred = X_b.dot(self.theta)
            # Hitung error
            errors = y_pred - y
            # Hitung gradien
            grad = (2 / m_samples) * X_b.T.dot(errors)
            # Update parameter theta
            self.theta -= self.lr * grad
            # Hitung nilai loss (MSE) dan simpan
            mse = np.mean(errors ** 2)
            self.loss_history.append(mse)

        return self

# 2. Daftar learning rate yang diuji
learning_rates = [0.001, 0.02, 0.1]
results = {}

print("--- HASIL PELATIHAN GRADIENT DESCENT ---")
for lr in learning_rates:
    gd_model = LinearRegressionGD(n_iterations=5000, lr=lr)
    gd_model.fit(X_train, y_train)

    results[lr] = {
        'theta': gd_model.theta,
        'loss_history': gd_model.loss_history
    }

    b_gd = gd_model.theta[0]
    m_gd = gd_model.theta[1]
    print(f"LR = {lr} -> Intercept (b): {b_gd:.4f} | Koefisien (m): {m_gd:.4f}")

# 3. Membuat Grafik Loss (MSE) vs Iterasi (menggunakan skala log y)
plt.figure(figsize=(10, 6))
for lr in learning_rates:
    plt.plot(results[lr]['loss_history'], label=f'LR = {lr}')

plt.xlabel('Jumlah Iterasi')
plt.ylabel('Loss (MSE - Skala Log)')
plt.title('Kurva Loss vs Iterasi untuk Berbagai Learning Rate')
plt.legend()
plt.grid(True, which="both", ls="--")
plt.show()
````

**Jawaban A5 (analisis):**
Eksperimen Gradient Descent menunjukkan bahwa learning rate α=0.02 merupakan nilai yang ideal karena berhasil membawa loss (MSE) menuju konvergensi secara cepat dan stabil hingga menghasilkan nilai parameter (b≈28.0323, m≈9.1402) yang serupa dengan metode Normal Equation (Soal A3). Sebaliknya, learning rate α=0.001 bekerja terlalu lambat sehingga proses penurunan loss masih sangat landai dan belum mencapai titik minimum optimal dalam 5000 iterasi. Sementara itu, learning rate α=0.1 mengalami kegagalan konvergensi (divergen) yang ditandai dengan peringatan overflow, karena ukuran langkah pembaharuan bobot terlalu besar sehingga algoritma terus melompati lembah minimum hingga nilainya membesar tak terhingga (infinity).

### Soal B2 – Split Data & Training
1. Bagi data dengan `test_size=0.25` dan `random_state=RANDOM_STATE`.
2. Latih model `LogisticRegression` (gradient descent, dari materi) dengan `learning_rate=0.1` dan `n_iterations=10000`.
3. Tampilkan intercept ($b$) dan koefisien ($m_1, m_2$), lalu tuliskan persamaannya:
   $$P(\text{Spam}) = \sigma(b + m_1 \cdot \text{JumlahLink} + m_2 \cdot \text{KataPromo})$$
4. Buat grafik **log loss vs iterasi**. Apakah loss sudah turun dan stabil?

````python
# Soal B2 - Poin 1: Split Data
from sklearn.model_selection import train_test_split

X_train_b2, X_test_b2, y_train_b2, y_test_b2 = train_test_split(
    X_email, y_email, test_size=0.25, random_state=5
)

print("Shape X_train_b2:", X_train_b2.shape)
print("Shape X_test_b2 :", X_test_b2.shape)
print("Shape y_train_b2:", y_train_b2.shape)
print("Shape y_test_b2 :", y_test_b2.shape)
````

````python
# Soal B2 - Poin 2: Training Model Logistic Regression (Gradient Descent)

# Fungsi Sigmoid
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

class LogisticRegressionGD:
    def __init__(self, n_iterations=10000, lr=0.1):
        self.n_iterations = n_iterations
        self.lr = lr
        self.theta = None
        self.loss_history = []

    def fit(self, X, y):
        m_samples = len(X)
        X_b = np.c_[np.ones((m_samples, 1)), X]
        self.theta = np.zeros(X_b.shape[1])
        self.loss_history = []

        for i in range(self.n_iterations):
            z = X_b.dot(self.theta)
            y_pred = sigmoid(z)
            grad = (1 / m_samples) * X_b.T.dot(y_pred - y)
            self.theta -= self.lr * grad

            epsilon = 1e-15
            y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
            loss = -np.mean(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
            self.loss_history.append(loss)

        return self

# Latih model
log_model = LogisticRegressionGD(n_iterations=10000, lr=0.1)
log_model.fit(X_train_b2, y_train_b2)
print("Model Logistic Regression berhasil dilatih!")
````

````python
# Soal B2 - Poin 3: Tampilkan Intercept, Koefisien, dan Persamaan
b_log = log_model.theta[0]
m1_log = log_model.theta[1]
m2_log = log_model.theta[2]

print(f"Intercept (b)      : {b_log:.4f}")
print(f"Koefisien m1 (Link): {m1_log:.4f}")
print(f"Koefisien m2 (Promo): {m2_log:.4f}")
print(f"\nPersamaan Sigmoid  : P(Spam) = sigma({b_log:.4f} + {m1_log:.4f} * JumlahLink + {m2_log:.4f} * KataPromo)")
````

````python
# Soal B2 - Poin 4: Grafik Log Loss vs Iterasi
plt.figure(figsize=(8, 5))
plt.plot(log_model.loss_history, color='purple')
plt.xlabel('Jumlah Iterasi')
plt.ylabel('Log Loss (Binary Cross-Entropy)')
plt.title('Kurva Log Loss vs Iterasi (Logistic Regression)')
plt.grid(True)
plt.show()
````

````python
# Jawaban Soal B3 - Evaluasi Model (Accuracy, Confusion Matrix, Precision, Recall, F1-Score)

# 1. Prediksi pada data testing (X_test_b2)
X_test_b2_b = np.c_[np.ones((len(X_test_b2), 1)), X_test_b2]
y_test_prob = sigmoid(X_test_b2_b.dot(log_model.theta))
y_pred_b3 = (y_test_prob >= 0.5).astype(int)

# 2. Hitung Confusion Matrix manual (TP, TN, FP, FN)
tp = np.sum((y_test_b2 == 1) & (y_pred_b3 == 1))
tn = np.sum((y_test_b2 == 0) & (y_pred_b3 == 0))
fp = np.sum((y_test_b2 == 0) & (y_pred_b3 == 1))
fn = np.sum((y_test_b2 == 1) & (y_pred_b3 == 0))

confusion_mat = np.array([[tn, fp], [fn, tp]])

# 3. Hitung Accuracy, Precision, Recall, dan F1-Score manual sesuai rumus
accuracy = (tp + tn) / len(y_test_b2)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print("--- HASIL EVALUASI MODEL (DATA TESTING) ---")
print(f"Confusion Matrix (TN, FP / FN, TP):\n{confusion_mat}\n")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1_score:.4f}")
````

**Jawaban B3 (analisis):**Dalam konteks filter spam, nilai FP (False Positive) terjadi ketika email normal (bukan spam) secara keliru diklasifikasikan oleh model sebagai spam, sedangkan FN (False Negative) terjadi ketika email spam berbahaya justru lolos dan masuk ke kotak masuk utama (inbox) sebagai email normal. Menurut pengguna, FP jauh lebih merugikan karena dapat menyebabkan email penting yang bersifat darurat atau krusial (seperti urusan pekerjaan, tugas kuliah, atau informasi akun penting) masuk ke folder spam dan terlewatkan tanpa disadari, sementara email spam yang masuk karena FN umumnya hanya berupa gangguan kecil yang masih bisa diabaikan secara manual.

**Jawaban B4 (analisis):**Visualisasi decision boundary dengan garis batas probabilitas 0,5 (P=0.5) memperlihatkan bagaimana model regresi logistik memisahkan ruang fitur antara kelas email spam dan bukan spam menggunakan background probabilitas berwarna gradasi. ### Soal B5 – Eksperimen Threshold
Secara default, model memprediksi Spam jika $P(\text{Spam}) \geq 0.5$.

1. Gunakan `predict_proba` lalu buat prediksi dengan threshold **0.3**, **0.5**, dan **0.7**.
2. Untuk setiap threshold, hitung confusion matrix, precision, dan recall. Tampilkan dalam satu tabel.
3. Jelaskan pengaruh menaikkan/menurunkan threshold terhadap precision dan recall. Berdasarkan jawaban Anda di Soal B3, **threshold mana yang Anda pilih** untuk filter spam? Beri alasan.

````python
# Jawaban Soal B5 - Eksperimen Threshold (0.3, 0.5, dan 0.7)
import pandas as pd

# 1. Hitung probabilitas data testing
X_test_b2_b = np.c_[np.ones((len(X_test_b2), 1)), X_test_b2]
y_test_prob = sigmoid(X_test_b2_b.dot(log_model.theta))

# Daftar threshold yang akan diuji
thresholds = [0.3, 0.5, 0.7]
eval_results = []

for th in thresholds:
    # Buat prediksi berdasarkan threshold tertentu
    y_pred_th = (y_test_prob >= th).astype(int)

    # Hitung Confusion Matrix manual
    tp = np.sum((y_test_b2 == 1) & (y_pred_th == 1))
    tn = np.sum((y_test_b2 == 0) & (y_pred_th == 0))
    fp = np.sum((y_test_b2 == 0) & (y_pred_th == 1))
    fn = np.sum((y_test_b2 == 1) & (y_pred_th == 0))

    # Hitung precision dan recall
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    accuracy = (tp + tn) / len(y_test_b2)

    eval_results.append({
        'Threshold': th,
        'Accuracy': round(accuracy, 4),
        'Precision': round(precision, 4),
        'Recall': round(recall, 4),
        'TP': tp, 'TN': tn, 'FP': fp, 'FN': fn
    })

# 2. Tampilkan dalam bentuk tabel
df_eval = pd.DataFrame(eval_results)
print("--- TABEL EKSPERIMEN THRESHOLD ---")
display(df_eval[['Threshold', 'Accuracy', 'Precision', 'Recall', 'TP', 'FP', 'TN', 'FN']])
````

**Jawaban B5 (analisis):**Berdasarkan hasil eksperimen berbagai threshold, menaikkan nilai threshold (seperti ke 0,7) akan membuat model menjadi lebih ketat dalam melabeli email sebagai spam, sehingga nilai precision meningkat karena jumlah kesalahan false positive berkurang, namun nilai recall cenderung turun akibat lebih banyak email spam lolos (false negative). Sebaliknya, menurunkan threshold (seperti ke 0,3) akan memperbesar nilai recall tetapi memperbanyak kesalahan false positive. Berdasarkan analisis sebelumnya di mana kesalahan false positive (email penting masuk folder spam) jauh lebih merugikan pengguna, maka threshold yang paling ideal dan dipilih untuk filter spam adalah nilai yang lebih tinggi (seperti 0.7) guna melindungi email penting agar tidak salah masuk ke folder spam.
